```python
import nltk
from nltk.util import ngrams
from collections import Counter, defaultdict

nltk.download('punkt')
nltk.download('punkt_tab')

sentences = [
    "The quick brown fox jumps over the lazy dog",
    "The quick brown fox is quick"
]

unigram_counts = Counter()
bigram_counts = Counter()
trigram_counts = Counter()

bigram_context = defaultdict(int)
trigram_context = defaultdict(int)

for sentence in sentences:
    tokens = nltk.word_tokenize(sentence.lower())

    unigrams = list(ngrams(tokens, 1))
    bigrams = list(ngrams(tokens, 2))
    trigrams = list(ngrams(tokens, 3))

    unigram_counts.update(unigrams)
    bigram_counts.update(bigrams)
    trigram_counts.update(trigrams)

    for bigram in bigrams:
        bigram_context[bigram[0]] += 1

    for trigram in trigrams:
        trigram_context[trigram[:2]] += 1

print("Unigram Counts:")
print(unigram_counts)

print("\nBigram Counts:")
print(bigram_counts)

print("\nTrigram Counts:")
print(trigram_counts)

print("\nUnigram Probabilities:")
total_unigrams = sum(unigram_counts.values())

for unigram, count in unigram_counts.items():
    print(unigram, ":", count / total_unigrams)

print("\nBigram Probabilities:")

for bigram, count in bigram_counts.items():
    probability = count / bigram_context[bigram[0]]
    print(bigram, ":", probability)

print("\nTrigram Probabilities:")

for trigram, count in trigram_counts.items():
    probability = count / trigram_context[trigram[:2]]
    print(trigram, ":", probability)
```

```python
def minimum_edit_distance(str1, str2):
    m, n = len(str1), len(str2)

    dp = [[0 for _ in range(n + 1)] for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i

    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if str1[i - 1] == str2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(
                    dp[i - 1][j],
                    dp[i][j - 1],
                    dp[i - 1][j - 1]
                )

    return dp[m][n], dp


def display_edit_distance_table(dp, str1, str2):
    print("\nEdit Distance Table:")
    print("  " + " ".join("#" + c for c in " " + str2))

    for i, row in enumerate(dp):
        prefix = "#" if i == 0 else str1[i - 1]
        print(f"{prefix} {' '.join(map(str, row))}")


def test_med():
    test_cases = [
        ("kitten", "sitting"),
        ("flaw", "lawn"),
        ("intention", "execution"),
        ("apple", "aple"),
        ("", "hello"),
        ("hello", "")
    ]

    for str1, str2 in test_cases:
        print(f"\nComparing: '{str1}' -> '{str2}'")
        med, dp = minimum_edit_distance(str1, str2)
        print(f"Minimum Edit Distance: {med}")
        display_edit_distance_table(dp, str1, str2)


test_med()
```

```python
class Parser:
    def __init__(self, grammar, start_symbol="S"):
        self.grammar = grammar
        self.start_symbol = start_symbol
        self.rules = self._build_rules(grammar)

    def _build_rules(self, grammar):
        rules = {}
        for lhs, rhs in grammar:
            if lhs not in rules:
                rules[lhs] = []
            rules[lhs].append(rhs)
        return rules

    def top_down_parse(self, sentence):
        success, tree, index = self._top_down_helper(
            self.start_symbol, sentence, 0
        )
        if success and index == len(sentence):
            return tree
        return None

    def _top_down_helper(self, symbol, sentence, index):
        if symbol not in self.rules:
            if index < len(sentence) and sentence[index] == symbol:
                return True, (symbol,), index + 1
            return False, None, index

        for production in self.rules[symbol]:
            children = []
            new_index = index
            success = True

            for sym in production:
                success, subtree, new_index = self._top_down_helper(
                    sym, sentence, new_index
                )
                if not success:
                    break
                children.append(subtree)

            if success:
                return True, (symbol, children), new_index

        return False, None, index

    def bottom_up_parse(self, sentence):
        stack = []
        input_buffer = sentence.copy()
        steps = []

        while True:
            if stack == [self.start_symbol] and not input_buffer:
                steps.append(("ACCEPT", stack.copy(), input_buffer.copy()))
                return steps

            if input_buffer:
                word = input_buffer.pop(0)
                stack.append(word)
                steps.append(("SHIFT", stack.copy(), input_buffer.copy()))

            reduced = True
            while reduced:
                reduced = False
                for lhs, rhs_list in self.rules.items():
                    for rhs in rhs_list:
                        if len(stack) >= len(rhs) and stack[-len(rhs):] == rhs:
                            stack = stack[:-len(rhs)] + [lhs]
                            steps.append((
                                f"REDUCE {lhs}->{rhs}",
                                stack.copy(),
                                input_buffer.copy()
                            ))
                            reduced = True
                            break
                    if reduced:
                        break

            if not input_buffer and stack != [self.start_symbol]:
                steps.append(("REJECT", stack.copy(), input_buffer.copy()))
                return steps


grammar = [
    ("S", ["NP", "VP"]),
    ("NP", ["Det", "N"]),
    ("VP", ["V", "NP"]),
    ("VP", ["V"]),
    ("Det", ["a"]),
    ("Det", ["the"]),
    ("N", ["cat"]),
    ("N", ["dog"]),
    ("V", ["chases"]),
    ("V", ["eats"])
]

parser = Parser(grammar)

sentence = ["the", "cat", "chases", "a", "dog"]

top_down_result = parser.top_down_parse(sentence)
print("Top-Down Parse Tree:")
print(top_down_result)

print("\nBottom-Up Parsing Steps:")
steps = parser.bottom_up_parse(sentence)

for action, stack, buffer in steps:
    print(f"{action:15} | Stack: {stack} | Buffer: {buffer}")
```
